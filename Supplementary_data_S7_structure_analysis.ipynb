Each AlphaFold model was structurally aligned to its reference using TM-align, which reports the TM-score (normalized independently
by query and reference length), root-mean-square deviation (RMSD), aligned length, and a residue-level sequence alignment. A TM-score > 0.5 was interpreted as indicating the same fold. 

```python
#tmalign_to_csv.py
#!/usr/bin/env python3
"""
Parse TM-align output files (-outfmt 1) into a summary CSV.

Usage:
    python tmalign_to_csv.py <folder_of_tmalign_txt> <output.csv>

Each TM-align .txt file must be named  <QUERY>_vs_<REFERENCE>.txt
e.g.  FN1326_CpsA_vs_1F75.txt

The script extracts, for every pairwise alignment:
  - query / reference names and lengths
  - aligned length, RMSD, sequence identity over the aligned region
  - TM-score normalized by the query length and by the reference length
  - reference coverage (% of the reference that is aligned)
"""
import sys, os, re, glob, csv

def parse_tmalign(path):
    txt = open(path).read()
    base = os.path.basename(path).replace('.txt', '')
    # split into query and reference on the FIRST '_vs_'
    query, ref = base.split('_vs_', 1)

    # The two header lines start with '>' and carry L= and TM-score=
    heads = re.findall(r'>(\S+)\s+L=(\d+)\s+d0=[\d.]+\s+seqID=[\d.]+\s+TM-score=([\d.]+)', txt)
    # heads[0] = structure_1 (query), heads[1] = structure_2 (reference)
    Lq, TMq = int(heads[0][1]), float(heads[0][2])
    Lr, TMr = int(heads[1][1]), float(heads[1][2])

    # The summary line:  # Lali=208  RMSD=1.49  seqID_ali=0.442
    m = re.search(r'Lali=(\d+)\s+RMSD=\s*([\d.]+)\s+seqID_ali=([\d.]+)', txt)
    Lali, RMSD, seqid = int(m.group(1)), float(m.group(2)), float(m.group(3))

    cov = round(Lali / Lr * 100, 1)   # % of reference covered
    return [query, ref, Lq, Lr, Lali, RMSD, round(seqid, 3), round(TMq, 3), round(TMr, 3), cov]

def main():
    folder, outcsv = sys.argv[1], sys.argv[2]
    rows = [parse_tmalign(f) for f in sorted(glob.glob(os.path.join(folder, '*.txt')))]
    with open(outcsv, 'w', newline='') as fh:
        w = csv.writer(fh)
        w.writerow(['Query', 'Reference', 'Query_len', 'Ref_len', 'Aligned_len',
                    'RMSD_A', 'Seq_ID_aligned', 'TM_score_norm_query',
                    'TM_score_norm_ref', 'Ref_coverage_pct'])
        w.writerows(rows)
    print(f'Wrote {len(rows)} rows to {outcsv}')

if __name__ == '__main__':
    main()

```

```python
#mafft_identity_matrix.py
#!/usr/bin/env python3
"""
Compute a pairwise percent-identity matrix from a MAFFT alignment and save it as CSV.

Usage:
    mafft --auto all_inhibitors.fasta > all_inhibitors.aln
    python mafft_identity_matrix.py all_inhibitors.aln inhibitor_identity_matrix.csv

Percent identity between two sequences =
    (number of alignment columns where both residues are identical and non-gap)
    / (number of columns where BOTH sequences have a residue, i.e. non-gap in both)
    * 100
"""
import sys, csv
from Bio import AlignIO

def pct_id(s1, s2):
    both = ident = 0
    for a, b in zip(s1, s2):
        if a != '-' and b != '-':
            both += 1
            if a == b:
                ident += 1
    return round(100.0 * ident / both, 1) if both else 0.0

def main():
    aln = AlignIO.read(sys.argv[1], 'fasta')
    names = [r.id for r in aln]
    seqs = [str(r.seq) for r in aln]
    n = len(names)
    matrix = [[pct_id(seqs[i], seqs[j]) for j in range(n)] for i in range(n)]
    with open(sys.argv[2], 'w', newline='') as fh:
        w = csv.writer(fh)
        w.writerow([''] + names)
        for name, row in zip(names, matrix):
            w.writerow([name] + row)
    print(f'Wrote {n}x{n} identity matrix to {sys.argv[2]}')

if __name__ == '__main__':
    main()
```

```python
#catalytic_residue_mapping.py
#!/usr/bin/env python3
"""
catalytic_residue_mapping.py

Parses TM-align (-outfmt 1) output files and maps known catalytic/functional
residues from reference proteins onto the query (AlphaFold model) sequences,
producing the catalytic_residue_conservation.csv table.

HOW IT WORKS:
  1. For each protein pair, the script reads the TM-align sequence alignment
     (query line + reference line, with gaps as '-').
  2. For each known catalytic motif in the reference (defined below), it finds
     the motif in the reference alignment sequence, identifies the alignment
     columns of the key residues, and checks what residues the query has at
     those same columns.
  3. It reports whether the motif/residues are conserved and at the same
     structural position (same alignment column).

  This motif-based approach is robust to leading-residue stripping by TMalign
  and to PDB numbering offsets, because it searches for the motif string in
  the alignment rather than using fixed residue numbers.

USAGE:
  python catalytic_residue_mapping.py <tmalign_output_dir> <output.csv>

  The directory should contain .txt files named <QUERY>_vs_<REFERENCE>.txt
  (as produced by run_all_analyses.sh or by running TMalign manually).
"""

import os
import sys
import re
import csv

CATALYTIC_MOTIFS = [
    # --- FN1326 CpsA vs 1F75 (UppS) ---
    # UppS has two Asp-rich Mg2+-binding motifs: MDGNGR (D at offset 1) and MPDPEL (D at offset 2)
    {
        "query": "FN1326_CpsA",
        "ref": "1F75",
        "features": [
            ("1st Asp-rich (Mg2+)", r"MDGNGR", [1],
             "D29 (MDGNGR)", "Catalytic Asp, Mg2+ binding"),
            ("2nd Asp-rich (Mg2+)", r"MPDPEL", [2],
             "D191 (MPDPEL)", "Catalytic Asp, Mg2+ binding"),
        ],
    },
    # --- FN1325 CpsB vs 4Q2G (CdsA) ---
    # CAP family catalytic motifs: KDS, GHGG, DRID
    {
        "query": "FN1325_CpsB",
        "ref": "4Q2G",
        "features": [
            ("CAP KDS motif", r"KDS", [0, 1],
             "K252/D253", "CDP-DAG synthase catalytic"),
            ("CAP GHGG motif", r"GHGG", [0, 1, 2, 3],
             "G260-H261-G262-G263", "Catalytic His (H261)"),
            ("CAP DRID motif", r"DRID", [0, 1, 2, 3],
             "D266-R267-I268-D269", "Catalytic Asp; I→F conservative"),
        ],
    },
    # --- FN1322 MucP vs 7W6X (RseP) ---
    # S2P zinc-metalloprotease HExGH motif: H and H are Zn ligands, E is catalytic
    {
        "query": "FN1322_MucP",
        "ref": "7W6X",
        "features": [
            ("HExGH zinc motif", r"HE.GH", [0, 1, 3, 4],
             "H22/E23/G25/H26", "Zn ligands (H,H) + catalytic Glu"),
        ],
    },
    # --- FN1885 Hemolysin III vs 5LXG (AdipoR1/PAQR) ---
    # PAQR Zn-binding His cluster: H251...H255...H265 in 5LXG
    # Match string: HQIFHVLVVAAAFVH → H at offsets 0, 4, 14
    {
        "query": "FN1885_HemolysinIII",
        "ref": "5LXG",
        "features": [
            ("C-term His cluster", r"HQIFHVLVVAAAFVH", [0, 4, 14],
             "H337/H341/H351", "PAQR Zn-binding His"),
        ],
    },
    # --- FN1318 SigA vs 3IYD chain F (sigma70) ---
    # Region 4.2 HPSR motif + core RNAP-binding block
    # Note: query has VVSIAKK (V→I conservative), ref has VISIAKK
    {
        "query": "FN1318_SigA",
        "ref": "ref_3IYD_F",
        "features": [
            ("Region 4.2 (-35 recog)", r"HPSR", [0, 1, 2, 3],
             "HPSR", "-35 promoter recognition"),
            ("Core RNAP-binding (reg 2.2/3.0)", "BLOCK:VISIAKK:ADQ:VVSIAKK", [],
             "VISIAKK...ADQ (60aa)", "52/60 identical (87%)"),
        ],
    },
    # --- FN1321 PilR vs 1NY5 (NtrC) ---
    # Walker A P-loop: GESGVGKE — key residues G, E, G, K (positions 0,1,3,6)
    {
        "query": "FN1321_PilR",
        "ref": "1NY5",
        "features": [
            ("Walker A (P-loop)", r"GESGVGKE", [0, 1, 3, 6],
             "GESGVGKE (G167-K174)", "ATPase P-loop, G/E/G/K conserved"),
        ],
    },
    # --- FN0845 HEPN vs 4NQF (HEPN domain) ---
    # HEPN catalytic motif: R-X4-6-H. In 4NQF: RTAELH at seq 124 (last match)
    # In query: RNQIVH at seq 25
    # Use general pattern; find_motif_in_aligned takes first match in query,
    # last match in ref (via LAST_MATCH prefix)
    {
        "query": "FN0845_HEPN",
        "ref": "4NQF",
        "features": [
            ("R-X4-6-H motif", r"R.{4,6}H", [0, 5],
             "R127/H132 (RTAELH)",
             "HEPN catalytic motif; weak structural align (TM<0.5); motif-based ID"),
        ],
    },
    # --- FN0846 YwqK vs 1H3I (SET7/9) ---
    # SET-domain catalytic Tyr in GYDH motif
    {
        "query": "FN0846_YwqK",
        "ref": "1H3I",
        "features": [
            ("Catalytic Tyr", r"GYDH", [1],
             "Y337 (GYDH)", "TM=0.34<0.5; NOT a SET fold (antitoxin)"),
        ],
    },
]


# ============================================================================
# PARSING AND MAPPING FUNCTIONS
# ============================================================================

def parse_tmalign_outfmt1(filepath):
    """Parse a TM-align -outfmt 1 output file.
    Returns (query_aligned_seq, ref_aligned_seq, stats_dict)."""
    with open(filepath) as f:
        lines = f.readlines()
    if len(lines) < 5:
        return None, None, {}
    query_aln = lines[1].rstrip()
    ref_aln = lines[3].rstrip()
    stats = {}
    for part in lines[4].strip().split("\t"):
        part = part.strip().lstrip("#")
        if "=" in part:
            k, v = part.split("=", 1)
            stats[k.strip()] = v.strip()
    return query_aln, ref_aln, stats


def build_position_map(aligned_seq):
    """Build map: alignment_column → residue_number (1-based). Gaps excluded."""
    pos_map = {}
    res_num = 0
    for col, char in enumerate(aligned_seq):
        if char != "-":
            res_num += 1
            pos_map[col] = res_num
    return pos_map


def find_motif_in_aligned(aligned_seq, motif_regex, take_last=False):
    """Find a motif in the gap-stripped sequence.
    Returns (match_obj, list_of_alignment_columns_for_each_residue_in_match).
    If take_last=True, returns the last match instead of the first."""
    clean_seq = aligned_seq.replace("-", "")
    matches = list(re.finditer(motif_regex, clean_seq))
    if not matches:
        return None, []
    match = matches[-1] if take_last else matches[0]

    # Map each residue in the match back to alignment columns
    pos_map = build_position_map(aligned_seq)
    match_start_res = match.start() + 1  # 1-based residue number

    cols = []
    for offset in range(len(match.group())):
        target_res = match_start_res + offset
        col = None
        for c, num in pos_map.items():
            if num == target_res:
                col = c
                break
        cols.append(col)

    return match, cols


def assess_residue_conservation(ref_char, qry_char):
    """Check if ref→qry is a conservative substitution."""
    if qry_char is None or qry_char == "-":
        return "No", "gap"
    if ref_char is None:
        return "No", "ref not found"
    if ref_char == qry_char:
        return "Yes", "identical"
    conservative = {
        ("D", "E"), ("E", "D"),
        ("K", "R"), ("R", "K"),
        ("S", "T"), ("T", "S"),
        ("I", "V"), ("V", "I"), ("I", "L"), ("L", "I"), ("V", "L"), ("L", "V"),
        ("F", "Y"), ("Y", "F"), ("F", "L"), ("L", "F"), ("I", "F"), ("F", "I"),
        ("A", "G"), ("G", "A"),
    }
    if (ref_char, qry_char) in conservative:
        return "Yes", "conservative"
    return "No", f"{ref_char}→{qry_char}"


def process_block_identity(query_aln, ref_aln, start_motif, end_motif, qry_start_alt=None):
    """Find a block from start_motif to end_motif in both sequences.
    If qry_start_alt is provided, use it as the start motif for the query
    (handles conservative substitutions like V→I)."""
    qry_seq = query_aln.replace("-", "")
    ref_seq = ref_aln.replace("-", "")

    # Try the main start motif, then the alternative for the query
    qry_start_pos = qry_seq.find(start_motif)
    if qry_start_pos < 0 and qry_start_alt:
        qry_start_pos = qry_seq.find(qry_start_alt)
    ref_start_pos = ref_seq.find(start_motif)
    if qry_start_pos < 0 or ref_start_pos < 0:
        return None, None, 0, 0

    qry_end = qry_seq.find(end_motif, qry_start_pos + len(start_motif))
    ref_end = ref_seq.find(end_motif, ref_start_pos + len(start_motif))
    if qry_end < 0 or ref_end < 0:
        return None, None, 0, 0

    qry_block = qry_seq[qry_start_pos:qry_end + len(end_motif)]
    ref_block = ref_seq[ref_start_pos:ref_end + len(end_motif)]

    min_len = min(len(qry_block), len(ref_block))
    if min_len == 0:
        return qry_block, ref_block, 0, 0
    identical = sum(1 for a, b in zip(qry_block, ref_block) if a == b)
    return qry_block, ref_block, identical, min_len


def process_pair(entry, tmalign_dir):
    """Process one protein pair and return CSV rows."""
    query = entry["query"]
    ref = entry["ref"]

    # Find TMalign output file
    filepath = None
    for f in os.listdir(tmalign_dir):
        if f.startswith(f"{query}_vs_{ref}") and f.endswith(".txt"):
            filepath = os.path.join(tmalign_dir, f)
            break
    if filepath is None:
        print(f"  WARNING: No TMalign output for {query} vs {ref}")
        return []

    query_aln, ref_aln, stats = parse_tmalign_outfmt1(filepath)
    if query_aln is None:
        print(f"  WARNING: Could not parse {filepath}")
        return []

    qry_map = build_position_map(query_aln)
    ref_map = build_position_map(ref_aln)

    rows = []
    for feature_name, motif_spec, key_offsets, ref_display, notes in entry["features"]:
        # Special case: block identity
        if isinstance(motif_spec, str) and motif_spec.startswith("BLOCK:"):
            parts = motif_spec.split(":")
            start_motif = parts[1]
            end_motif = parts[2]
            qry_start_alt = parts[3] if len(parts) > 3 else None
            qry_block, ref_block, ident, total = process_block_identity(
                query_aln, ref_aln, start_motif, end_motif, qry_start_alt)
            if qry_block and ref_block:
                pct = 100 * ident / total if total > 0 else 0
                ref_str = f"{ref_block[:10]}...{ref_block[-6:]} ({len(ref_block)}aa)"
                qry_str = f"{qry_block[:10]}...{qry_block[-6:]} ({len(qry_block)}aa)"
                conserved = f"Yes ({pct:.0f}%)" if pct >= 50 else "No"
                aligned = "Yes"
            else:
                ref_str = "block not found"
                qry_str = "block not found"
                conserved = "No"
                aligned = "No"
            rows.append({
                "Query": query, "Reference": ref,
                "Functional_feature": feature_name,
                "Ref_residue": ref_display, "Query_residue": qry_str,
                "Conserved": conserved, "Aligned_same_position": aligned,
                "Notes": notes,
            })
            continue

        # Motif-based mapping
        # For HEPN (R-X4-6-H), take the last match in the reference (canonical RTAELH)
        ref_take_last = "HEPN" in query
        ref_match, ref_cols = find_motif_in_aligned(ref_aln, motif_spec, take_last=ref_take_last)
        qry_match, qry_cols = find_motif_in_aligned(query_aln, motif_spec)

        if ref_match is None:
            ref_str = "motif not found in ref"
            qry_str = qry_match.group() if qry_match else "not found"
            conserved = "No"
            aligned = "No"
        elif qry_match is None:
            # Motif not found in query — check what's at the same alignment columns
            ref_residues = []
            qry_residues = []
            all_conserved = True
            all_aligned = True
            for offset in key_offsets:
                col = ref_cols[offset] if offset < len(ref_cols) else None
                if col is not None:
                    ref_char = ref_aln[col]
                    ref_res_num = ref_map.get(col, "?")
                    ref_residues.append(f"{ref_char}{ref_res_num}")
                    qry_char = query_aln[col]
                    qry_res_num = qry_map.get(col, "gap")
                    if qry_char == "-":
                        qry_residues.append("gap")
                        all_conserved = False
                        all_aligned = False
                    else:
                        qry_residues.append(f"{qry_char}{qry_res_num}")
                        cons, _ = assess_residue_conservation(ref_char, qry_char)
                        if cons == "No":
                            all_conserved = False
                else:
                    ref_residues.append("?")
                    all_aligned = False

            ref_str = "/".join(ref_residues)
            qry_str = "/".join(qry_residues)
            conserved = "Yes" if all_conserved else "No"
            aligned = "Yes" if all_aligned else "No"
        else:
            # Both have the motif — check key residues at aligned positions
            ref_residues = []
            qry_residues = []
            all_conserved = True
            all_aligned = True

            # First check: does the query motif align to the same region as ref?
            qry_start_col = qry_cols[0] if qry_cols else None
            ref_start_col = ref_cols[0] if ref_cols else None
            same_region = (qry_start_col is not None and ref_start_col is not None
                           and abs(qry_start_col - ref_start_col) <= 10)

            if same_region:
                # Motifs are in the same structural region — report query motif directly
                for offset in key_offsets:
                    if offset < len(qry_cols):
                        col = qry_cols[offset]
                        if col is not None:
                            qry_char = query_aln[col]
                            qry_res_num = qry_map.get(col, "?")
                            qry_residues.append(f"{qry_char}{qry_res_num}")
                        else:
                            qry_residues.append("?")
                    else:
                        qry_residues.append("?")

                for offset in key_offsets:
                    if offset < len(ref_cols):
                        col = ref_cols[offset]
                        if col is not None:
                            ref_char = ref_aln[col]
                            ref_res_num = ref_map.get(col, "?")
                            ref_residues.append(f"{ref_char}{ref_res_num}")
                        else:
                            ref_residues.append("?")
                    else:
                        ref_residues.append("?")

                # Check conservation by comparing motif residues directly
                for offset in key_offsets:
                    if offset < len(qry_cols) and offset < len(ref_cols):
                        qry_col = qry_cols[offset]
                        ref_col = ref_cols[offset]
                        if qry_col is not None and ref_col is not None:
                            cons, _ = assess_residue_conservation(
                                ref_aln[ref_col], query_aln[qry_col])
                            if cons == "No":
                                all_conserved = False
            else:
                # Motifs are in different regions — report both motifs separately
                # The query has the motif but it doesn't align to the ref motif
                # (weak structural alignment, e.g. HEPN TM<0.5)
                for offset in key_offsets:
                    if offset < len(ref_cols):
                        col = ref_cols[offset]
                        if col is not None:
                            ref_char = ref_aln[col]
                            ref_res_num = ref_map.get(col, "?")
                            ref_residues.append(f"{ref_char}{ref_res_num}")
                        else:
                            ref_residues.append("?")
                    else:
                        ref_residues.append("?")

                    if offset < len(qry_cols):
                        col = qry_cols[offset]
                        if col is not None:
                            qry_char = query_aln[col]
                            qry_res_num = qry_map.get(col, "?")
                            qry_residues.append(f"{qry_char}{qry_res_num}")
                        else:
                            qry_residues.append("?")
                    else:
                        qry_residues.append("?")

                # Check if key residues match (R and H for HEPN)
                for offset in key_offsets:
                    if offset < len(qry_cols) and offset < len(ref_cols):
                        qry_col = qry_cols[offset]
                        ref_col = ref_cols[offset]
                        if qry_col is not None and ref_col is not None:
                            cons, _ = assess_residue_conservation(
                                ref_aln[ref_col], query_aln[qry_col])
                            if cons == "No":
                                all_conserved = False
                all_aligned = False

            ref_str = "/".join(ref_residues)
            qry_str = "/".join(qry_residues)
            conserved = "Yes" if all_conserved else "No"
            aligned = "Yes" if all_aligned else "No"

        rows.append({
            "Query": query, "Reference": ref,
            "Functional_feature": feature_name,
            "Ref_residue": ref_display, "Query_residue": qry_str,
            "Conserved": conserved, "Aligned_same_position": aligned,
            "Notes": notes,
        })

    return rows


def main():
    if len(sys.argv) < 3:
        print("Usage: python catalytic_residue_mapping.py <tmalign_dir> <output.csv>")
        print()
        print("  tmalign_dir: folder containing TM-align -outfmt 1 .txt files")
        print("  output.csv:  path for the output CSV table")
        sys.exit(1)

    tmalign_dir = sys.argv[1]
    output_csv = sys.argv[2]

    all_rows = []
    for entry in CATALYTIC_MOTIFS:
        print(f"  Processing {entry['query']} vs {entry['ref']}...")
        rows = process_pair(entry, tmalign_dir)
        all_rows.extend(rows)

    fieldnames = ["Query", "Reference", "Functional_feature", "Ref_residue",
                  "Query_residue", "Conserved", "Aligned_same_position", "Notes"]
    with open(output_csv, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_rows)

    print(f"\nWrote {len(all_rows)} rows to {output_csv}")


if __name__ == "__main__":
    main()
```

```python
# pLDDT-colored
# adding color to the picture in PyMOL.
#colored according to pLDDT
# not PDB color, only changes the appearance inside PyMOL for the rendered image.

# if keep not work, run it with bash, copy the code to a .py file and run it with bash.
# I think is PyMOL is getting stuck on startup

import pymol
from pymol import cmd
import glob, os

pymol.finish_launching(['pymol','-qc'])
cmd.bg_color('white')
cmd.set('ray_shadows', 0)
cmd.set('antialias', 2)
cmd.set('ray_trace_mode', 0) #ray tracing 1 , high-quality images really slow.

# pLDDT coloring: blue(high) -> white -> red(low), like AlphaFold
# Use spectrum: red (<50) orange (70) yellow (90) blue (>90) - AF convention
def color_plddt(sel):
    cmd.color('0xD9573B', f'{sel} and b < 50')          # very low - red
    cmd.color('0xF0A13A', f'{sel} and b > 49.9 and b < 70')  # low - orange
    cmd.color('0xF5E663', f'{sel} and b > 69.9 and b < 90')  # confident - yellow
    cmd.color('0x3B6FB6', f'{sel} and b > 89.9')          # very high - blue

models = [
 ('FN1326_CpsA','CpsA / UppS'),
 ('FN1325_CpsB','CpsB / CdsA'),
 ('FN1322_MucP','MucP'),
 ('FN1885_HemolysinIII','Hemolysin III'),
 ('FN1318_SigA','SigA'),
 ('FN1321_PilR','PilR'),
 ('FN0845_HEPN','HEPN'),
 ('FN0846_YwqK','YwqK'),
]
os.makedirs('figs/plddt_panels', exist_ok=True)
for pdb,label in models:
    cmd.delete('all')
    cmd.load(f'models_pdb/{pdb}.pdb','mol')
    cmd.hide('everything')
    cmd.show('cartoon','mol')
    color_plddt('mol')
    cmd.orient('mol')
    cmd.zoom('mol', 2)
    cmd.png(f'figs/plddt_panels/{pdb}.png', width=900, height=900, dpi=150, ray=0) # make ray=0 so much faster
    print('rendered', pdb)
cmd.quit()
```

```python
# run the same analysis, print out result.
# FN1321_PilR.pdb is the sequence to alphfold, dowload cif and convert pdb
# naming pilr, but we not sure if it is pilr, so we need to check the structure and compare with atoC
import os
import re
import subprocess

# ============================================================
# Settings
# ============================================================
TMALIGN_BIN = './TMalign'   # IMPORTANT: use local executable
QUERY_LABEL = 'FN1321_PilR'

comparisons = [
    ('FN1321_PilR.pdb', '1NY5.pdb',                  'FN1321_PilR_vs_1NY5.txt'),
    ('FN1321_PilR.pdb', '1OJL_A.pdb',                'FN1321_PilR_vs_1OJL_A.txt'),
    ('FN1321_PilR.pdb', 'AF-Q06065-F1-model_v6.pdb', 'FN1321_PilR_vs_AF_AtoC.txt'),
    ('FN1321_PilR.pdb', 'AF-Q00934-F1-model_v6.pdb', 'FN1321_PilR_vs_AF_PilR.txt'),
]

REF_NAMES = {
    '1NY5':    '1NY5 (NtrC1)',
    '1OJL_A':  '1OJL (ZraR)',
    'AF_AtoC': 'AF-Q06065 (AtoC)',
    'AF_PilR': 'AF-Q00934 (PilR)',
}

FILE_KEYS = {
    'FN1321_PilR_vs_1NY5.txt':        '1NY5',
    'FN1321_PilR_vs_1OJL_A.txt':      '1OJL_A',
    'FN1321_PilR_vs_AF_AtoC.txt':     'AF_AtoC',
    'FN1321_PilR_vs_AF_PilR.txt':     'AF_PilR',
}

os.makedirs('tmalign_outputs', exist_ok=True)

# ============================================================
# Step 1: Run TM-align
# ============================================================
for model, ref, outfile in comparisons:
    outpath = f'tmalign_outputs/{outfile}'

    # If file exists and is non-empty, skip
    if os.path.exists(outpath) and os.path.getsize(outpath) > 0:
        print(f"Exists, skipping: {outfile}")
        continue

    cmd = [TMALIGN_BIN, f'models_pdb/{model}', f'refs/{ref}', '-outfmt', '1']
    print(f"Running: {model} vs {ref}")

    with open(outpath, 'w') as fh:
        subprocess.run(cmd, stdout=fh, stderr=subprocess.PIPE, text=True, check=False)

    if os.path.getsize(outpath) == 0:
        print(f"  WARNING: {outfile} is empty (TM-align may have failed)")
    else:
        print(f"  Wrote: {outfile}")

# ============================================================
# Step 2: Parsing helpers
# ============================================================
def parse_tmalign(path):
    with open(path) as f:
        text = f.read().strip()

    if not text:
        raise ValueError(f"{path} is empty.")

    # Grab all TM-score lines
    tm_scores = re.findall(r'TM-score=\s*([0-9.]+)', text)
    lengths = re.findall(r'\bL=\s*(\d+)', text)

    # Parse key summary line
    m = re.search(
        r'Lali=(\d+)\s+RMSD=\s*([0-9.]+)\s+seqID_ali=([0-9.]+)',
        text
    )

    if len(tm_scores) < 2:
        raise ValueError(f"Could not find two TM-scores in {path}")
    if len(lengths) < 2:
        raise ValueError(f"Could not find two sequence lengths in {path}")
    if not m:
        raise ValueError(f"Could not parse Lali/RMSD/seqID_ali in {path}")

    Lq = int(lengths[0])
    Lr = int(lengths[1])
    TMq = float(tm_scores[0])
    TMr = float(tm_scores[1])
    Lali = int(m.group(1))
    RMSD = float(m.group(2))
    seqid = float(m.group(3))
    cov = round((Lali / Lr) * 100, 1) if Lr else 0.0

    return {
        'Lq': Lq,
        'Lr': Lr,
        'Lali': Lali,
        'RMSD': round(RMSD, 2),
        'seqid': round(seqid, 3),
        'TMq': round(TMq, 3),
        'TMr': round(TMr, 3),
        'cov': cov,
    }

# ============================================================
# Step 3: Parse results and print table
# Requested order:
# Query  Reference  Query_len  Ref_len  Aligned_len  RMSD_A
# Seq_ID_aligned  TM_score_norm_query  TM_score_norm_ref  Ref_coverage_pct
# ============================================================
rows = []

for filename, ref_key in sorted(FILE_KEYS.items()):
    filepath = f'tmalign_outputs/{filename}'
    if not os.path.exists(filepath):
        print(f"WARNING: {filepath} not found, skipping")
        continue

    try:
        s = parse_tmalign(filepath)
    except Exception as e:
        print(f"ERROR parsing {filepath}: {e}")
        continue

    rows.append([
        QUERY_LABEL,
        REF_NAMES[ref_key],
        s['Lq'],
        s['Lr'],
        s['Lali'],
        s['RMSD'],
        s['seqid'],
        s['TMq'],
        s['TMr'],
        s['cov']
    ])

# ============================================================
# Step 4: Print table
# ============================================================
print("\n" + "=" * 120)
print(
    f"{'Query':<18} {'Reference':<24} {'Query_len':>9} {'Ref_len':>8} "
    f"{'Aligned_len':>12} {'RMSD_A':>8} {'Seq_ID_aligned':>15} "
    f"{'TM_score_norm_query':>20} {'TM_score_norm_ref':>18} {'Ref_coverage_pct':>18}"
)
print("=" * 120)

for row in rows:
    print(
        f"{row[0]:<18} {row[1]:<24} {row[2]:>9} {row[3]:>8} "
        f"{row[4]:>12} {row[5]:>8.2f} {row[6]:>15.3f} "
        f"{row[7]:>20.3f} {row[8]:>18.3f} {row[9]:>18.1f}"
    )

print("=" * 120)
```

```python

# NOTE: For 1OJL (ZraR), PDB residue numbers start at 140 (it is a fragment,
# residues 141-441 of full-length ZraR). The build_pos_map function counts
# from 1 in the aligned sequence, so computed ref_str gives wrong numbers
# for 1OJL (e.g. position 30 instead of PDB residue 169).
# Fix: use hardcoded REF_LABELS for the Ref_residue field (see below).
# The computed ref_str is only used for verification.
# ============================================================================
# Walker A motif conservation — FN1321 (PilR/AtoC) vs 4 sigma-54 bEBP refs
# ============================================================================
# Reads TM-align -outfmt 1 output from Cell 19 (tmalign_outputs/FN1321_PilR_vs_*.txt)
# and checks Walker A P-loop conservation across all 4 references.
#
# File names from Cell 19 use "FN1321_PilR" (the original annotation).
# Display label uses "FN1321_AtoC" (the validated identity).

import os
import re

# ============================================================
# Settings — must match Cell 19's file naming
# ============================================================
TMALIGN_DIR = "tmalign_outputs"          
WALKER_A_REGEX = r"G.{4}GK[DEST]"        # General Walker A: G-x(4)-GK-[D/E/S/T]

# ============================================================
# Comparisons
# query_file = file lookup name 
# query_display = output table label (validated identity = AtoC)
# ============================================================
CATALYTIC_MOTIFS = [
    {
        "query_file": "FN1321_PilR",        
        "query_display": "FN1321_PilR",     
        "ref": "1NY5",
        "ref_display": "1NY5 (NtrC1)",
        "features": [
            ("Walker A (P-loop)", WALKER_A_REGEX, [0, 1, 3, 6],
             "GESGVGKE (G167-E174)",
             "AAA+ ATPase P-loop; compare FN1321 (AtoC) to NtrC1"),
        ],
    },
    {
        "query_file": "FN1321_PilR",
        "query_display": "FN1321_PilR",
        "ref": "1OJL_A",
        "ref_display": "1OJL (ZraR)",
        "features": [
            ("Walker A (P-loop)", WALKER_A_REGEX, [0, 1, 3, 6],
             "GDSGTGKE (G169-E176)",
             "AAA+ ATPase P-loop; compare FN1321 (AtoC) to ZraR"),
        ],
    },
    {
        "query_file": "FN1321_PilR",
        "query_display": "FN1321_PilR",
        "ref": "AF_AtoC",
        "ref_display": "AF-Q06065 (AtoC)",
        "features": [
            ("Walker A (P-loop)", WALKER_A_REGEX, [0, 1, 3, 6],
             "GESGTGKE (G173-E180)",
             "AAA+ ATPase P-loop; compare FN1321 (AtoC) to E. coli AtoC"),
        ],
    },
    {
        "query_file": "FN1321_PilR",
        "query_display": "FN1321_PilR",
        "ref": "AF_PilR",
        "ref_display": "AF-Q00934 (PilR)",
        "features": [
            ("Walker A (P-loop)", WALKER_A_REGEX, [0, 1, 3, 6],
             "GESGSGKE (G163-E170)",
             "AAA+ ATPase P-loop; compare FN1321 (AtoC) to P. aeruginosa PilR"),
        ],
    },
]

# ============================================================
# Helpers
# ============================================================

def parse_tmalign_outfmt1(filepath):
    """
    Parse TM-align -outfmt 1 output.
    Returns: query_aln, ref_aln, stats_dict
    """
    with open(filepath) as f:
        lines = f.readlines()

    if len(lines) < 5:
        return None, None, {}

    query_aln = lines[1].rstrip()
    ref_aln = lines[3].rstrip()

    stats = {}
    for part in lines[4].strip().split("\t"):
        part = part.strip().lstrip("#")
        if "=" in part:
            k, v = part.split("=", 1)
            stats[k.strip()] = v.strip()

    return query_aln, ref_aln, stats


def build_pos_map(aligned_seq):
    """Map alignment column -> residue number (1-based)."""
    pos_map = {}
    res_num = 0
    for col, char in enumerate(aligned_seq):
        if char != "-":
            res_num += 1
            pos_map[col] = res_num
    return pos_map


def find_motif(aligned_seq, motif_regex):
    """
    Find first motif match in gap-stripped aligned sequence.
    Returns: match_obj, list_of_alignment_columns
    """
    clean_seq = aligned_seq.replace("-", "")
    matches = list(re.finditer(motif_regex, clean_seq))
    if not matches:
        return None, []

    match = matches[0]
    pos_map = build_pos_map(aligned_seq)
    match_start_res = match.start() + 1

    cols = []
    for offset in range(len(match.group())):
        target_res = match_start_res + offset
        col = None
        for c, num in pos_map.items():
            if num == target_res:
                col = c
                break
        cols.append(col)

    return match, cols


def check_walker_a(query_aln, ref_aln):
    """
    Check Walker A motif conservation.
    Uses key_offsets [0, 1, 3, 6] = G, E, G, K (matches existing S13 format).

    Returns: ref_str, qry_str, conserved, aligned
    """
    qry_match, qry_cols = find_motif(query_aln, WALKER_A_REGEX)
    ref_match, ref_cols = find_motif(ref_aln, WALKER_A_REGEX)

    qry_map = build_pos_map(query_aln)
    ref_map = build_pos_map(ref_aln)

    key_offsets = [0, 1, 3, 6]  # G, E, G, K

    if ref_match is None:
        return "motif not found", "not found", "No", "No"

    ref_res = []
    qry_res = []
    all_cons = True
    all_ali = True

    for off in key_offsets:
        if off < len(ref_cols) and ref_cols[off] is not None:
            col = ref_cols[off]
            rc = ref_aln[col]
            rn = ref_map.get(col, "?")
            ref_res.append(f"{rc}{rn}")

            if col < len(query_aln):
                qc = query_aln[col]
                if qc == "-":
                    qry_res.append("gap")
                    all_cons = False
                    all_ali = False
                else:
                    qn = qry_map.get(col, "?")
                    qry_res.append(f"{qc}{qn}")
                    if rc != qc:
                        cons_pairs = {("D", "E"), ("E", "D"), ("S", "T"), ("T", "S")}
                        if (rc, qc) not in cons_pairs:
                            all_cons = False
            else:
                qry_res.append("?")
                all_ali = False
        else:
            ref_res.append("?")
            qry_res.append("?")
            all_ali = False

    ref_str = f"{ref_match.group()} ({ref_res[0]}-{ref_res[-1]})"
    qry_str = "/".join(qry_res)

    return ref_str, qry_str, "Yes" if all_cons else "No", "Yes" if all_ali else "No"


def process_pair(entry, tmalign_dir):
    """Process one comparison: find TM-align file, parse, check Walker A."""
    query_file = entry["query_file"]       # for file lookup
    query_display = entry["query_display"]  # for output table
    ref = entry["ref"]
    ref_display = entry["ref_display"]

    # Find TM-align output file — uses query_file (FN1321_PilR) to match Cell 19
    filepath = None
    for f in os.listdir(tmalign_dir):
        if f.startswith(f"{query_file}_vs_{ref}") and f.endswith(".txt"):
            filepath = os.path.join(tmalign_dir, f)
            break

    if filepath is None:
        print(f"  WARNING: No TM-align output for {query_file} vs {ref} in {tmalign_dir}")
        return []

    query_aln, ref_aln, stats = parse_tmalign_outfmt1(filepath)
    if query_aln is None:
        print(f"  WARNING: Could not parse {filepath}")
        return []

    rows = []
    for feature_name, motif_regex, key_offsets, ref_label, notes in entry["features"]:
        ref_str_computed, qry_str, cons, ali = check_walker_a(query_aln, ref_aln)

        rows.append({
            "Query": query_display,           # display validated identity
            "Reference": ref_display,
            "Functional_feature": feature_name,
            "Ref_residue": ref_label,         # hardcoded (correct PDB residue numbers)
            "Query_residue": qry_str,         # computed from alignment
            "Conserved": cons,
            "Aligned_same_position": ali,
            "Notes": notes,
        })

        print(f"  {ref_display}:")
        print(f"    ref_label        = {ref_label}")
        print(f"    ref_str_computed  = {ref_str_computed}")
        print(f"    query_residue     = {qry_str}")
        print(f"    conserved         = {cons}")
        print(f"    aligned           = {ali}")

    return rows


# ============================================================
# Main
# ============================================================

if not os.path.isdir(TMALIGN_DIR):
    print(f"ERROR: TM-align directory '{TMALIGN_DIR}' not found.")
    print("Run Cell 19 first to generate TM-align outputs.")
else:
    print(f"Using TM-align directory: {TMALIGN_DIR}\n")

    all_rows = []
    for entry in CATALYTIC_MOTIFS:
        print(f"Processing {entry['query_file']} vs {entry['ref']} ...")
        rows = process_pair(entry, TMALIGN_DIR)
        all_rows.extend(rows)

    # Print summary table
    print("\n" + "=" * 130)
    print(f"{'Query':<18} {'Reference':<22} {'Feature':<22} {'Ref_residue':<28} {'Query_residue':<22} {'Cons':<6} {'Align':<6}")
    print("=" * 130)

    for row in all_rows:
        print(
            f"{row['Query']:<18} {row['Reference']:<22} {row['Functional_feature']:<22} "
            f"{row['Ref_residue']:<28} {row['Query_residue']:<22} "
            f"{row['Conserved']:<6} {row['Aligned_same_position']:<6}"
        )

    print("=" * 130)

    # Also write CSV for S13 manual entry
    if all_rows:
        import csv
        csv_path = "atoC_validation_WalkerA.csv"
        with open(csv_path, "w", newline="") as f:
            w = csv.writer(f)
            w.writerow(["Query", "Reference", "Functional_feature", "Ref_residue",
                        "Query_residue", "Conserved", "Aligned_same_position", "Notes"])
            for row in all_rows:
                w.writerow([row["Query"], row["Reference"], row["Functional_feature"],
                            row["Ref_residue"], row["Query_residue"],
                            row["Conserved"], row["Aligned_same_position"], row["Notes"]])
        print(f"\nWrote {len(all_rows)} rows to {csv_path}")
```
